```python
from datasets import load_dataset

dataset = load_dataset("issai/kazparc", "kazparc")
```

```python
# extracting kazakh text from the dataset
def extract_kazakh_text(dataset, source_field="source_lang", target_field="target_lang"):
    kazakh_train = []
    kazakh_val = []
    kazakh_test = []

    for row in dataset["train"]:
        if row["pair"] == "kk_en":
            kazakh_train.append(row[source_field])
        elif row["pair"] == "en_kk":
            kazakh_train.append(row[target_field])

    for row in dataset["validation"]:
        if row["pair"] == "kk_en":
            kazakh_val.append(row[source_field])
        elif row["pair"] == "en_kk":
            kazakh_val.append(row[target_field])

    for row in dataset["test"]:
        if row["pair"] == "kk_en":
            kazakh_test.append(row[source_field])
        elif row["pair"] == "en_kk":
            kazakh_test.append(row[target_field])

    return kazakh_train, kazakh_val, kazakh_test

kazakh_train, kazakh_val, kazakh_test = extract_kazakh_text(dataset)

print(f"Sample Train Kazakh Texts: {kazakh_train[:5]}")
print(f"Sample Validation Kazakh Texts: {kazakh_val[:5]}")
print(f"Sample Test Kazakh Texts: {kazakh_test[:5]}")

with open("kazakh_train.txt", "w", encoding="utf-8") as f:
    for line in kazakh_train:
        f.write(line + "\n")

with open("kazakh_val.txt", "w", encoding="utf-8") as f:
    for line in kazakh_val:
        f.write(line + "\n")

with open("kazakh_test.txt", "w", encoding="utf-8") as f:
    for line in kazakh_test:
        f.write(line + "\n")

# stats
print(f"Total Train Kazakh Sentences: {len(kazakh_train)}")
print(f"Total Validation Kazakh Sentences: {len(kazakh_val)}")
print(f"Total Test Kazakh Sentences: {len(kazakh_test)}")
```
